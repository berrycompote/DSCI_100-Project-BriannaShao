#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
wmm_basketball <- wmm_basketball_raw |> # No need for creating new variables or transforming data into tidy form
    select(year, school, tourney_finish, reg_wins_pct, total_wins_pct, seed) |> # To make new data frame only contain year and school plus predictor and response variables
    filter(seed != 0) |> # Removed 8 rows where seed is not actual seed since there is no data on what the seed of the teams actually was
    mutate(tourney_finish = fct_recode(tourney_finish, "8" = "opening_round_loss",
                                                       "7" = "first_round_loss",
                                                       "6" = "second_round_loss",
                                                       "5" = "top_16_loss",
                                                       "4" = "top_8_loss",
                                                       "3" = "top_4_loss",
                                                       "2" = "top_2_loss",
                                                       "1" = "champ")) |> # Attempting to represent tourney_finish categories w/ numeric variables that can be used to calculate distance
    mutate(tourney_finish = as.numeric(tourney_finish)) |>
    mutate(seed = as_factor(seed)) |> # Converted seed variable into factor
    mutate(year = as_factor(year)) |>
    arrange(seed)
# mutate(tourney_finish = fct_recode(...)) argument results in warning because all the rows w/ tourney_finish as "opening_round_loss" were removed using filter()
    # "8" = "opening_round_loss" was kept in code just in case tidy data is changed to include the 8 rows w/ seed as 0 that were filtered out
head(wmm_basketball, 6)
```

```r
# Distribution of reg_wins_pct
options(repr.plot.width = 8, repr.plot.height = 6)
reg_wins_pct_dist <- wmm_basketball |>
    ggplot(aes(x = reg_wins_pct)) +
        geom_histogram(binwidth = 2.5) +
        labs(x = "Regional wins (%)", y = "Number of teams") +
        ggtitle("Distribution of Regional Win Percentage") +
        theme(text = element_text(size = 15))
reg_wins_pct_dist 
# Initially attempted to use facet_grid to make multiple subplots based on year/school/year+school, but there were too many levels in both to make a legible plot
```

```r
# Relative proportions of seed categories

seed_prop <- wmm_basketball |>
    group_by(seed) |>
    summarize(count = n()) |>
    mutate(prop = count / sum(count))

options(repr.plot.width = 8, repr.plot.height = 6)
seed_prop_plot <- seed_prop |>
    ggplot(aes(x = seed, y = prop)) +
        geom_bar(stat = "identity") +
        labs(x = "Seeding designation", y = "Proportion of teams") +
        ggtitle("Relative Proportions of Seeding Designations") +
        scale_y_continuous(limits = c(0, 0.085)) +
        theme(text = element_text(size = 15))
seed_prop_plot
```

Visualization 1: Plotting the relationship between reg_wins_pct and total_wins_pct allows us to pre-emptively gauge how distance calculations affecting these two predictor variables might affect the seeding designation of a new observation. We can also use this visualization to plot new observations and see a visual representation of the distance calculation.

```r
reg_wins_total_wins <- wmm_basketball |>
    ggplot(aes(x = reg_wins_pct, y = total_wins_pct, color = seed)) +
        geom_point(alpha = 0.4) +
        labs(x = "Regional wins (%)", y = "Total wins (%)", color = "Seeding designation") +
        ggtitle("Total Win Percentage vs. Regional Win Percentage") +
        theme(text = element_text(size = 15))
reg_wins_total_wins
```
